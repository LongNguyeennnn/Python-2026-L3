```python
student_name = []
cs = []
marks = {}
```

## Input function

```python
def input_student():
    num = int(input("Number of student: "))

    for i in range (num):
        student = {
            'name': input("Student name: "), 
            'id': input("Student ID: "), 
            'dob': input("DOB: ")
            }
        student_name.append(student)
```

```python
def input_course():
    csnum = int(input("Number of course: "))

    for j in range (csnum):
        course = {
            'name': input("Course name: "), 
            'id': input("Course ID: ")}
        cs.append(course)
```

```python
def input_mark():
    selected = input("ID of the course: ")
    check = any(c["id"] == selected for c in cs)
    if not check:
        print("The course does not exist")
        return

    if selected not in mark:
        mark[selected] = {}

    for _ in student_name:
        mark = float(input(f"Student mark {student_name['name']} - ID: {student_name['id']}: "))
        marks[selected][student_name["id"]] = mark
```

## Listing Functions

```python
def list_course():
    for i in cs:
        print(i['name'] + ' - ' + i['id'])
```

```python
def list_student():
    for j in student_name:
        print("Name: " + j['name'] + " - ID: " + j['id'] + ' - DOB: ' + j['dob'])
```

```python
def show_student_marks():
    if not marks:
        print("No marks entered yet.")
        return
    list_course()

    course_id = input("\nEnter course ID to show marks for: ")
    
    if course_id not in marks:
        print("No marks recorded for this course ID.")
        return

    print(f"\nMarks for Course ID: {course_id}")
    for student in student_name:
        s_id = student['id']
        mark = marks[course_id].get(s_id, "N/A")
        print(f"Student: {student['name']} (ID: {s_id}) -> Mark: {mark}")
```

## Main

```python
def main():
    while True:
        print("1. Input students")
        print("2. Input courses")
        print("3. Input marks for a course")
        print("4. List courses")
        print("5. List students")
        print("6. Show student marks for a given course")
        print("7. Exit")

        choice = input("Your choice: ")
        if choice == '1':
            input_student()
        elif choice == '2':
            input_course()
        elif choice == '3':
            input_mark()
        elif choice == '4':
            list_course()
        elif choice == '5':
            list_student()
        elif choice == '6':
            show_student_marks()
        elif choice == '7':
            print("Exiting program.")
            break
        else:
            print("Invalid choice, please try again.")


main()
```
