## Ex 3

```python
def prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

c = int(input("Enter a number? "))
if prime(c):
    print(f"{c} is a prime number.")
else:
    print(f"{c} is not a prime number.")
```

## Ex 4

```python
def perfect(n):
    sum = 1
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            sum += i
            if i != n // i:
                sum += n // i
    if sum == n:
        return True
    return False

d = int(input("Enter a number? "))
if perfect(d):
    print(f"{d} is a perfect number.")
else:
    print(f"{d} is not a perfect number.")
```

## Ex 9

```python
def factorial(n):
    if n < 1:
        return -1
    product = 1
    for i in range (1, n+1):
        product = product * i
    return product

u = int(input())
print(factorial(u))
```

## Ex 10

```python
def divisor(n):
    sum = []
    if n <= 1:
        sum.append(1)
        return sum
    else:
        for i in range(1, int(n**0.5) + 1):
            if n % i == 0:
                sum.append(i)
                if i != n // i:
                    sum.append(n // i)
        sum.sort()
        return sum

x = int(input())
print(divisor(x))
```
